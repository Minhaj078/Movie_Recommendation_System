```python
import ast

def convert(obj):
    L = []
    for i in ast.literal_eval(obj):
         ## ast.literal_eval() is used to safely evaluate a string containing a Python literal or container display. It can be used to convert a string representation of a list, dictionary, tuple, or other Python data structure into the corresponding Python object.
        L.append(i['name'])
    return L

movies['genres'] = movies['genres'].apply(convert)
movies.head()
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
movies.head()
```

```python
import ast

def convert1(obj):
    L = []
    counter = 0
    for i in ast.literal_eval(obj):
        if counter < 3:  # Only take the first 3 genres
            L.append(i['name'])
            counter += 1
    return L

movies['cast'] = movies['cast'].apply(convert1)
movies.head()
```

```python
def fetch_director(obj):
    L = []
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            L.append(i['name'])
            break  # Assuming there's only one director, we can break after finding it
    return L

movies['crew'] = movies['crew'].apply(fetch_director)

movies.head()
```

```python
movies['overview'] = movies['overview'].apply(lambda x: x.split())
movies['overview'].head()
```

```python
movies['generes'] = movies['genres'].apply(lambda x: [i.replace(" ", "") for i in x])
movies['keywords'] = movies['keywords'].apply(lambda x: [i.replace(" ", "") for i in x])
movies['cast'] = movies['cast'].apply(lambda x: [i.replace(" ", "") for i in x])
movies['crew'] = movies['crew'].apply(lambda x: [i.replace(" ", "") for i in x])

movies.head()
```

```python
new_df = movies[['movie_id','title','tags']].copy()
```

 NOTE:
 Porter Stemmer is used in this project because it is fast and works well
 for classical NLP techniques like CountVectorizer + Cosine Similarity.

 Limitation:
 Stemming simply removes word suffixes and may produce non-dictionary words
 (e.g., "studies" -> "studi", "running" -> "run").

 Better Alternative:
 Lemmatization (WordNetLemmatizer / spaCy) is generally preferred because
 it converts words to their actual dictionary form based on linguistic rules
 (e.g., "studies" -> "study", "better" -> "good"), resulting in more
 meaningful and accurate text processing.

 Future Improvement:
 Replace Porter Stemming with Lemmatization or Transformer-based embeddings
 (e.g., Sentence-BERT) to improve recommendation quality.

```python
import nltk
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()

# we use this library to reduce the words to their root form. For example, "running" will be reduced to "run". This helps in improving the accuracy of our recommendation system by treating different forms of a word as the same word.

def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)

```

```python
new_df['tags'] = new_df['tags'].apply(stem)
```

```python
new_df['tags'][1]
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000, stop_words='english')
vectors = cv.fit_transform(new_df['tags']).toarray()
```

```python
features = cv.get_feature_names_out().tolist()
print(features) 
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(vectors)
similarity[2]
```

```python
def recommend(movie):
    movie_index = new_df[new_df['title'] == movie].index[0]
    distances = similarity[movie_index]
    movies_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x: x[1])[1:6]

    for i in movies_list:
        print(new_df.iloc[i[0]].title)
```

```python
recommend("Iron Man")
```
